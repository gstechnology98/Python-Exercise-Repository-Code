<p style = "font-size:24px; font-weight:bold;">1. Write a Python program that asks the user to enter a number. If the input is not a valid integer, raise a <span style="background-color:yellow;">ValueError</span> and display a helpful error message instead of crashing.</p>

```python
try:
    user_input = input("Enter a number: ")
    number = int(user_input)
    print("You entered:", number)
except ValueError:
    print("Error: That was not a valid integer. Please enter a number.")
```

<p style="font-size:24px; font-weight:bold">2. Create a <span style = "color:blue;">function</span> <span style="background-color:yellow;">safe_divide(a, b)</span> that divides <span style="background-color:yellow;">a</span> by <span style="background-color:yellow;">b</span>. If <span style="background-color:yellow;">b</span> is zero, handle the <span style="background-color:yellow;">ZeroDivisionError</span> and return <span style="background-color:yellow;">None</span> instead of letting the program crash.

</p>

```python
def safe_divide(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        print("Warning: Division by zero is not allowed.")
        return None

print(safe_divide(10, 2))   # 5.0
print(safe_divide(10, 0))   # None
```
