<p style = "font-size:24px; font-weight:bold;">1. Write a script to perform the following three operations on given list
<br>i. Access the third element of a list
<br>ii. List Length: Print the total number of items
<br>iii. Check if the list is empty
</p>

```python
sample_list = [10, 20, 30, 40, 50, 60, 70, 80, 90]

# i) Access third Elements
print(f"Third element: {sample_list[2]}")

# b) List Length
print(f"Length of list: {len(sample_list)}")

# c) Check if Empty
is_empty = len(sample_list) == 0
print(f"Is the list empty? {is_empty}")
```

<p style = "font-size:24px; font-weight:bold;">2. Take a given list and modify it through five specific actions:
    <br>i. Change Element: Change the second element of a list to 200 and print the updated list.
    <br>ii. Append Element: Add 600 o the end of a list and print the new list.
    <br>iii. Insert Element: Insert 300 at the third position (index 2) of a list and print the result.
    <br>iv. Remove Element (by value): Remove 600 from the list and print the list.
    <br>v. Remove Element (by index): Remove the element at index 0 from the list print the list.
</p>

```python
list_m = [100, 50, 400, 500, 300, 80, 700]

# i) Change Element
list_m[1] = 200
print(f"Updated (Change): {list_m}")

# ii) Append Element
list_m.append(600)
print(f"Updated (Append) list: {list_m}")

# iii) Insert Element
list_m.insert(2, 300)
print(f"Updated (Insert) list: {list_m}")

# iv) Remove Element by value
list_m.remove(600)
print(f"Updated (Remove 600) from list: {list_m}")

# v) Remove Element by index
list_m.pop(0)
print(f"Updated (Remove Index 0): {list_m}")
```

<p style = "font-size:24px; font-weight:bold;">3. Calculate the total sum of all integers in a list and find the arithmetic mean (average).
</p>

```python
nums = [10, 20, 30, 40, 50]

total_sum = sum(nums)
average = total_sum / len(nums)

print(f"Sum: {total_sum}")
print(f"Average: {average}")
```

<p style = "font-size:24px; font-weight:bold;">4.  Identify the largest and smallest numerical values within a provided list.
</p>

```python
data_points = [45, 12, 89, 2, 67]

max_val = max(data_points)
min_val = min(data_points)

print(f"Maximum: {max_val}")
print(f"Minimum: {min_val}")
```
