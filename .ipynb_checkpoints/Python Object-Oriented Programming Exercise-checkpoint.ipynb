<p style = "font-size:20px; font-weight:bold;">2. Write a Python program to create a <span style="color:darkred;">Vehicle</span> class with two instance attributes: <span style="color:darkred;">max_speed </span> and <span style="color:darkred;">mileage</span>. Create an object of the class and print both attributes.</p>

```python
class Vehicle:
    def __init__(self, name, max_speed, mileage):
        self.name = name
        self.max_speed = max_speed
        self.mileage = mileage

vehicle1 = Vehicle("Tesla Model S", 250, 18)
print(f"Vehicle Name: {vehicle1.name}, Speed: {vehicle1.max_speed}, Mileage: {vehicle1.mileage}")
```

<p style = "font-size:20px; font-weight:bold;">3. Write a Python program to create a <span style="color:darkred;">Rectangle</span> class with <span style="color:darkred;">length</span> and <span style="color:darkred;">width</span> as instance attributes, and two methods: <span style="color:darkred;">area()</span> that returns the area and <span style="color:darkred;">perimeter()</span> that returns the perimeter.</p>

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

rect = Rectangle(10, 4)
print("Area =", rect.area())
print("Perimeter =", rect.perimeter())
```
