<p style="font-size:23px; font-weight:bold;">7. Given a list of numbers, iterate through it and print numbers that satisfy these conditions:<br>
a. The number must be divisible by five.<br>
b. If the number is greater than 150, skip it and move to the next.<br>
c. If the number is greater than 500, stop the loop entirely.
</p>

```python
numbers = [12, 75, 150, 180, 145, 525, 50]

for item in numbers:
    # Condition 3: Stop the loop if number > 500
    if item > 500:
        break
    
    # Condition 2: Skip the number if > 150
    if item > 150:
        continue
    
    # Condition 1: Print if divisible by 5
    if item % 5 == 0:
        print(item)
```

<p style="font-size:23px; font-weight:bold;">11. Write a program that takes a string and reverses it using a <b style="color:red;"> for</b> loop. While Python’s <b style="color:red;"> [::-1]</b> shortcut is famous, reversing a string manually is a classic way to understand how sequences are constructed.</p>

```python
original_str = "Python"
reversed_str = ""

for char in original_str:
    # Prepending: Put the new character before the existing ones
    reversed_str = char + reversed_str

print(f"Original: {original_str}")
print(f"Reversed: {reversed_str}")
```

<p style = "font-size:23px; font-weight:bold;">12. Write a program that counts the total number of vowels and consonants in a given sentence, ignoring spaces and special characters. </p>

```python
sentence = "Loops are Fun! It's helpful for running codes"
vowels = "aeiou"
v_count = 0
c_count = 0

for char in sentence.lower():
    if char.isalpha():  # Only process letters
        if char in vowels:
            v_count += 1
        else:
            c_count += 1

print(f"Vowels: {v_count}")
print(f"Consonants: {c_count}")
```

<p style = "font-size:23px; font-weight:bold">13. Write a program to count the total number of digits in a given integer using a while loop.</p>

```python
num = 75869
count = 0

# Keep dividing by 10 until the number reaches 0
while num != 0:
    # Floor division to remove the last digit
    num = num // 10
    # Increment the counter
    count = count + 1

print("Total digits are:", count)
```

<p style = "font-size:23px; font-weight:bold;">14. Write a program to reverse a given integer number (e.g., 76542 should become 24567).</p>

```python
num = 76542890
reverse_number = 0

while num > 0:
    # Get the last digit
    digit = num % 10
    # Add it to the reverse (shifting existing digits left)
    reverse_number = (reverse_number * 10) + digit
    # Remove the last digit from the original number
    num = num // 10

print(reverse_number)
```

<p style = "font-size:23px; font-weight:bold;">15. Write a program to find the largest and smallest digit within a given integer (e.g., in 75869, the largest is 9 and the smallest is 5).</p>

```python
num = 87934124056

# Initialize with opposite extremes
largest = 0
smallest = 9

while num>0:
    digit = num%10
    # Check for new largest
    if digit>largest:
        largest = digit

    # Check for new smallest
    if digit<smallest:
        smallest = digit

    num = num//10

print("Largest Digit is: ", largest)
print("Smallest Digit is: ", smallest)
```

<p style = "font-size:23px; font-weight:bold;">16. Write a program to check if a given number is a palindrome. A palindrome number is a number that remains the same when its digits are reversed (e.g., 121, 343).</p>

```python
number = 121
# Store original to compare later
temp = number
reverse_num = 0

while number > 0:
    digit = number % 10
    reverse_num = (reverse_num * 10) + digit
    number = number // 10

if temp == reverse_num:
    print(f"Yes. The {temp} number is palindrome number")
else:
    print(f"No. The {temp} number is not palindrome")
```

<p style = "font-size:23px; font-weight:bold;">17. Write a program to use a loop to find the factorial of a given number (e.g., 5!). The factorial of N is the product of all integers from 1 to N.</p>

```python
num = 5
factorial = 1

if num < 0:
    print("Factorial does not exist for negative numbers")
elif num == 0:
    print("The factorial of 0 is 1")
else:
    for i in range(1, num + 1):
        factorial = factorial * i
    print(f"The factorial of {num} is {factorial}")
```

<p style="font-size:23px; font-weight:bold;">18. The Collatz conjecture states that if you start with any positive integer n, and if n is even, divide it by 2; if n is odd, multiply it by 3 and add 1. Repeat the process. The sequence will always eventually reach 1. Write a program to print this sequence for a given number.</p>

```python
n=78
print(n, end="")

while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = (3 * n) + 1
    print(f", {n}", end="")
```

<p style = "font-size:23px; font-weight:bold;">19. Write a program to check if a number is an Armstrong number. An Armstrong number (for a 3-digit number) is an integer such that the sum of the cubes of its digits is equal to the number itself (e.g., 153 = 1^3 + 5^3 + 3^3).</p>

```python
num = 157
num_str = str(num)
power = len(num_str)
total = 0

for digit in num_str:
    total += int(digit) ** power

if total == num:
    print(f"{num} is an Armstrong number")
else:
    print(f"{num} is not an Armstrong number")
```

<p style = "font-size:23px; font-weight:bold;">23. Write a program to print a triangle pattern where each row consists of the same letter, and the letter changes (increments) with each new row.</p>

```python
rows = 5
ascii_value = 65 # Starting with 'A'

for i in range(rows):
    # Calculate current letter
    letter = chr(ascii_value + i)
    # Print the letter (i + 1) times
    for j in range(i + 1):
        print(letter, end=" ")
    print()
```

<p style="font-size:23px; font-weight:bold;">24. Print a 5*5 square of stars where the middle is empty, leaving only the border.</p>

```python
n = 5
for i in range(n):
    for j in range(n):
        # Check if we are on the border
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

<p style = "font-size:23px; font-weight:bold;">25.  Write a program to print the following pattern using nested loops:</p>

```python
rows = 5

#Upper part of pattern
for i in range(0, rows):
    for j in range(0, i+1):
        print("*", end="")
    print("\r")

#Lower part of pattern
for i in range(rows - 1, 0, -1):
    for j in range(0, i):
        print("*", end="")
    print("\r")
```

<p style = "font-size:24px; font-weight:bold;">31. Given a list of integers, move all even numbers to the beginning of the list and all odd numbers to the end.</p>

```python
numbers = [1, 2, 3, 4, 5, 6]
evens = []
odds = []

for n in numbers:
    if n % 2 == 0:
        evens.append(n)
    else:
        odds.append(n)

# Merging lists
result = evens + odds
print("Segregated List:", result)
```

<p style = "font-size:24px; font-weight:bold;">32. Given a list and an integer k, rotate the list to the left by k positions. For example, if k=2, the first two elements move to the end of the list.</p>

```python
nums = [1, 2, 3, 4, 5, 6, 7, 8, 9]
k = 2

# Perform the rotation k times
for _ in range(k):
    # Remove the first element
    first_element = nums.pop(0)
    # Move it to the end
    nums.append(first_element)

print("Rotated List:", nums)
```

<p style = "font-weight:bold; font-size:24px">33. Write a program to count the frequency of each word in a given string.</p>

```python
text = "apple banana apple orange banana apple"
words = text.split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        # First time seeing this word
        frequency[word] = 1

print(frequency)
```

<p style="font-size:24px; font-weight:bold">35. Write a program to check if a number is a “Perfect Number.” A perfect number is a positive integer that is equal to the sum of its proper divisors (excluding the number itself). For example, 6 is perfect because 1 + 2 + 3 = 6.</p>

```python
num = 28
divisor_sum = 0

# A divisor cannot be greater than half the number
for i in range(1, (num // 2) + 1):
    if num % i == 0:
        divisor_sum += i
        

if divisor_sum == num:
    print(f"{num} is a Perfect Number")
else:
    print(f"{num} is not a Perfect Number")
```

<p style = "font-size:24px; font-weight:bold;">36. Manually convert a binary string (e.g., <span style="background-color:yellow;">"1101"</span>) into its decimal integer equivalent using a loop. Do not use <span style="background-color:yellow;">int(binary, 2)</span>.</p>

```python
binary_str = "1101"
decimal_val = 0

# Reverse to handle indices as powers (0, 1, 2...)
reversed_binary = binary_str[::-1]

for i in range(len(reversed_binary)):
    if reversed_binary[i] == '1':
        decimal_val += 2 ** i

print("Decimal value:", decimal_val)
```

<p style="font-size:24px; font-weight:bold;">37. Write a program to display all prime numbers within a range (e.g., 25 to 50). A prime number is a natural number greater than 1 that is not a product of two smaller natural numbers.</p>

```python
start = 25
end = 50

print(f"Prime numbers between {start} and {end} are:")

for num in range(start, end + 1):
    # Prime numbers are greater than 1
    if num > 1:
        for i in range(2, num):
            if (num % i) == 0:
                # Found a factor, not prime
                break
        else:
            # Loop finished without finding a factor
            print(num)
```

<p style="font-size:24px;font-weight:bold;">38. Write a program to calculate the sum of the series 2 + 22 + 222 + 2222 + …. up to N terms. For example, if n=5, the series is 2 + 22 + 222 + 2222 + 22222.</p>

```python
n = 5
start = 2
sum_seq = 0

for i in range(0, n):
    # Add current term to the total sum
    sum_seq += start
    # Generate the next term (e.g., 2 -> 22 -> 222)
    start = start * 10 + 2

print(sum_seq)
```

<p style="font-size:24px; font-weight:bold;">39. Given a nested list (a list containing other lists), write a program to “flatten” it into a single list containing all the individual elements.</p>

```python
nested_list = [[10, 20], [30, 40], [50, 60]]
flattened = []

# Iterate through each sub-list
for sublist in nested_list:
    # Iterate through each item in the current sub-list
    for item in sublist:
        flattened.append(item)

print("Flattened List:", flattened)
```

<p style="font-size:24px; font-weight:bold;">40. Given a 2D list (matrix), find the row and column index of a target value.</p>

```python
matrix = [
    [10, 20],
    [30, 40],
    [50, 60]
]
target = 30

for r_idx, row in enumerate(matrix):
    for c_idx, val in enumerate(row):
        if val == target:
            print(f"Target {target} found at Row: {r_idx}, Column: {c_idx}")
            break # Found it!
```
