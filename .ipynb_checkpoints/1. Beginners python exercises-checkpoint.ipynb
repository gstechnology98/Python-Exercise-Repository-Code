<p style = "font-size:18px; font-weight:bold;">1. Write a Python function that accepts two integer numbers. If the product of the two numbers is less than or equal to 1000, return their product; otherwise, return their sum.</p>

```python
def prod_or_sum(a,b):
    n = a*b
    if n<=1000:
        return n
    else:
        return a+b

result= prod_or_sum(30,20)
print("The result (Product or sum) of a and b is : ",result)

result = prod_or_sum(40, 30)
print("The result (Product or sum) of a and b is : ",result)
```

<p style = "font-size:18px; font-weight:bold; ">2. Iterate through the first 10 numbers (0–9). In each iteration, print the current number, the previous number, and their sum.</p>

```python
print("Printing current and previous number sum in a range(10)")
prev_num = 0

for i in range(10):
    x_sum = prev_num+i
    print(f"Current number {i} Previous number {prev_num} Sum: {x_sum}")

    prev_num = i
```

<p style="font-size:20px; font-weight:bold">3. Display only those characters which are present at an even index number in given string.</p>

```python
word = "pynative"

print("Original String is ",word)

even_chars = word[0::2]

print("Printing only even index chars")

for n in even_chars:
    print(n)
```

<p style="font-size:20px; font-weight:bold">4. Write a function to remove characters from a string starting from index 0 up to n and return a new string.</p>

```python
def delete_chars(word, n):
    print('Original string:', word)
    # Extract string from index n to the end
    res = word[n:]
    return res

print("Removing characters from a string")
print(delete_chars("pynative", 4))
print(delete_chars("pynative", 2))
```

<p style="font-size:20px; font-weight:bold">5. Write a program to swap the values of two variables, a and b, without using a third temporary variable.</p>

```python
a = 6
b = 20

print(f"Before Swap: a = {a} and b = {b}")

a,b = b,a

print(f"After Swap: a = {a} and b = {b}")
```

<p style = "font-size:20px; font-weight:bold;">6. Write a program that calculates the factorial of a given number (e.g., 5!) using a for loop.</p>

```python
n = 8

fact  = 1

for i in range(1, n+1):
    fact = fact * i

print(f"The factorial of {n} is {fact}")
```

<p style="font-size:20px; font-weight:bold;">8. Write a program that takes a string and reverses it (e.g., “Python” becomes “nohtyP”).</p>

<p style="font-size:20px; font-weight:bold;"> 9. Write a program to count the total number of vowels (a, e, i, o, u) present in a given sentence.</p>

```python
user_input = "Python3 is a programming language"
contains_digit = False

# Iterate through each character
for char in user_input:
    if char.isdigit():
        contains_digit = True
        break  # Exit early since we found one

print(f"The string '{user_input}' contains digits: {contains_digit}")
```

<p style="font-size:23px; font-weight:bold;">36. Write a program to capitalize the first letter of each word in a given string without using the built-in .title() method.</p>

```python
text = "hello world from python program"

# Split the string into a list of words
words = text.split()
capitalized_words = []

for word in words:
    # Capitalize each word and add to the new list
    capitalized_words.append(word.capitalize())

# Join the list back into a single string
result = " ".join(capitalized_words)
print(result)
```

<p style="font-size:23px; font-weight:bold;">37. Create a countdown timer that starts from a given number and counts down to zero using a while loop.</p>

```python
import time
count = 10

while count>0:
    print(count)

     # Pause the program for 1 second
    time.sleep(1)
    
    # Decrement the counter
    count -=1

print("Blast off!")
```

<p style = "font-size:23px; font-weight:bold;">38. Write a program that creates a new text file named <b style="color:red;">notes.txt</b>, writes three separate lines of text to it, and then reads that file back to display the contents in the console.</p>

```python
# Part 1: Writing to the file
with open("notes.txt", "w") as file:
    file.write("Hello, this is my first note.\n")
    file.write("Python file handling is simple.\n")
    file.write("End of file.\n")

# Part 2: Reading from the file
print("Reading file contents:")
with open("notes.txt", "r") as file:
    content = file.read()
    print(content)
```

<p style="font-size: 23px; font-weight: bold;">39. Write a script that opens an existing <span style="color:red;">.txt</span> file and counts the total number of words it contains.</p>

```python
# Assuming 'sample.txt' exists with some text
try:
    with open("sample.txt", "r") as file:
        data = file.read()
        words = data.split()
        word_count = len(words)
        print(f"The file contains {word_count} words.")
except FileNotFoundError:
    print("Error: The file 'sample.txt' was not found.")
```

<p style = "font-size:23px; font-weight:bold;">40. Create a <span style="color:red;">Car</span> class with attributes for <span style="color:red;">make </span>, <span style="color:red;">model</span>, and <span style="color:red;">year</span>. Include a method called <span style="color:red;">start_engine()</span> that prints a formatted string describing the car starting up.
</p>

```python
class Car:
    def __init__(self, make, model, year):
        # Setting up attributes
        self.make = make
        self.model = model
        self.year = year

    def start_engine(self):
        # A method that uses the object's attributes
        print(f"The {self.year} {self.make} {self.model}'s engine is now running!")

# Creating an object (an instance of the class)
my_car = Car("Toyota", "Camry", 2022)

# Calling the method
my_car.start_engine()
```
