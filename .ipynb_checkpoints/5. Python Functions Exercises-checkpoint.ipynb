<p style = "font-size:24px; font-weight:bold;">2. Create a function <span style='color:red;'>func1()</span> such that it can accept a variable number of arguments and print all of them. Whether you pass two numbers or five, the function should handle them all without error.</p>

```python
def func1(*args):
    print("Printing values:")
    for i in args:
        print(i)

# Calling with 3 arguments
func1(20, 40, 60)

# Calling with 2 arguments
func1(80, 100)
```
