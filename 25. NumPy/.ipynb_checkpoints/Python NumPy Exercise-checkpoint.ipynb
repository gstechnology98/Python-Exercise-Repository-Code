<h2>3. Print Array Attributes</h2>
<p>Instructions: Print the following attributes of the array:</p>
<ul>
    <li>The shape of the array.</li>
    <li>The number of array dimensions.</li>
    <li>The size of each element in bytes.</li>
</ul>

```python
import numpy as np

# Create the 4x2 array of type unsigned int16
# We use np.array() and specify the dtype
my_array = np.array([[1, 2], [3, 4], [5, 6], [7, 8]], dtype=np.uint16)

# Print the attributes
print("The shape of the array is:", my_array.shape)
print("The number of dimensions is:", my_array.ndim)
print("The size of each element in bytes is:", my_array.itemsize)
```

<h2>13. Extract Odd Rows and Even Columns</h2>
<p>Extract an array of odd rows and even columns from below NumPy array.</p>


```python
import numpy as np

# Original array
sampleArray = np.array([
    [3, 6, 9, 12], 
    [15, 18, 21, 24], 
    [27, 30, 33, 36], 
    [39, 42, 45, 48], 
    [51, 54, 57, 60]
])

# Use slicing to select odd rows and even columns
# The slice [1::2] selects every second element starting from index 1 (odd rows).
# The slice [::2] selects every second element starting from index 0 (even columns).
newArray = sampleArray[1::2, ::2]

print("Original Array:")
print(sampleArray)

print("\nNew Array with Odd Rows and Even Columns:")
print(newArray)
```

<h2>16. Replace all odd numbers in a NumPy array with -1</h2>

```python
import numpy as np

arr = np.arange(1, 11)
print("Original Array:", arr)

arr[arr % 2 == 1] = -1
print("Modified Array:", arr)
```

<h2>18. Find the common items between two arrays</h2>

```python
import numpy as np

a = np.array([1, 2, 3, 2, 8, 4, 2, 4])
b = np.array([2, 4, 5, 6, 8])
common_items = np.intersect1d(a, b)
print(common_items)
```

<h1>21. Compute the mean, median, and standard deviation of a NumPy array</h1>
<ul>
    <li><b>Mean</b>: Average of numbers.</li>
    <li><b>Median</b>: Middle value when sorted.</li>
    <li><b>Standard Deviation</b>: How spread out values are from the mean.</li>
</ul>

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
```

<h1>22. Remove common items from array
</h1>

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5])
b = np.array([5, 6, 7, 8, 9])

# Create a boolean mask where `a`'s elements are NOT in `b`
# mask = np.in1d(a, b)
mask = np.isin(a,b)

# Use the `~` operator to invert the mask (get the opposite)
result = a[~mask]
print(result)
```

<h1>23. Normalize a NumPy array (values between 0 and 1)</h1>

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
normalized = (arr - arr.min()) / (arr.max() - arr.min())
print("Normalized Array:", normalized)
```

<h1>24. Get the positions where elements of array a and b match</h1>

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5])
b = np.array([1, 4, 3, 7, 8])

# Use np.where() to find the index where the values match
index = np.where(a == b)
print(index)
```

<h1>25. Extract numbers from an array</h1>
<p>Write a code to extract all numbers from an array that are between 5 and 10 (inclusive).</p>

```python
import numpy as np

arr = np.arange(15)
result = arr[(arr >= 5) & (arr <= 10)]
print(result)
```

<h1>27. Sorting a NumPy array based on a specific column</h1>

```python
import numpy as np

sampleArray = np.array([[34, 43, 73], [82, 22, 12], [53, 94, 66]])

# Get the indices that would sort the array by its second column
sorted_indices = sampleArray[:, 1].argsort()

# Use the sorted indices to reorder the rows of the array
sorted_array = sampleArray[sorted_indices]

print("Original array:")
print(sampleArray)

print("\nSorted array:")
print(sorted_array)
```
