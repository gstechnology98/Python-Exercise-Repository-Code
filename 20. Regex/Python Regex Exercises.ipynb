<p style = "font-size:24px; font-weight:bold;">1. Write a Python program to verify that a string contains only alphanumeric characters (a-z, A-Z, and 0-9).</p>

```python
import re

text = "Hello123"

if re.fullmatch(r"[a-zA-Z0-9]+", text):
    print("Valid: contains only alphanumeric characters")
else:
    print("Invalid: contains non-alphanumeric characters")
```

<p style = "font-weight:bold; font-size:24px">2. Write a Python program to match a string that has an <span style="background-color:#f509a3;">a</span> followed by zero or more <span style="background-color:#f509a3;">b</span>s (e.g., <span style="background-color:#f509a3;">a</span>, <span style="background-color:#f509a3;">ab</span>, <span style="background-color:#f509a3;">abb</span>). </p>


```python
import re
pattern = r"ab*"
test_strings = ["a", "ab", "abb", "abbb", "b", "ba","baa","baaa"]
for s in test_strings:
    result = re.fullmatch(pattern, s)
    print(f"{s:<6} -> {'Match' if result else 'No match'}")
```

<p style = "font-weight:bold; font-size:24px">3. Write a Python program to match a string that has an <span style="background-color:#f509a3;">a</span> followed by zero or more <span style="background-color:#f509a3;">b</span>s (e.g., <span style="background-color:#f509a3;">ab</span>, <span style="background-color:#f509a3;">abb</span>, but not <span style="background-color:#f509a3;">a</span>). </p>

```python
import re
pattern = r"ab*"
test_strings = ["a", "ab", "abb", "abbb", "b", "ba", "baa"]
for s in test_strings:
    result = re.fullmatch(pattern, s)
    print(f"{s:<6} -> {'Match' if result else 'No match'}")
```

<p style = "font-weight:bold; font-size:24px">4. Write a Python program to match a string that has an <span style="background-color:#f509a3;">a</span> followed by zero or one <span style="background-color:#f509a3;">b</span> (e.g., <span style="background-color:#f509a3;">a</span>, <span style="background-color:#f509a3;">ab</span>, but nothing else). </p>

```python
import re
pattern = r"ab?"
test_strings = ["a", "ab", "abb", "abbb", "b", "ba"]
for s in test_strings:
    result = re.fullmatch(pattern, s)
    print(f"{s:<6} -> {'Match' if result else 'No match'}")
```

<p style = "font-weight:bold; font-size:24px">5. Write a Python program to match a string that has an <span style="background-color:#f509a3;">a</span> followed by exactly three <span style="background-color:#f509a3;">b</span>s (e.g., only <span style="background-color:#f509a3;">abbbb</span>is a valid match). </p>

```python
import re
pattern = r"ab{3}"
test_strings = ["a", "ab", "abb", "abbb", "abbbb", "b"]
for s in test_strings:
    result = re.fullmatch(pattern, s)
    print(f"{s:<6} -> {'Match' if result else 'No match'}")
```
