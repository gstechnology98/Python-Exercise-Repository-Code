<p style = "font-weight:bold; font-size:24px;">1. Write a Python function that accepts two integer numbers. If the product of the two numbers is less than or equal to 1000, return their product; otherwise, return their sum.</p>

```python
def mult_or_sum(num1, num2):
    # Calculate product
    product = num1 * num2
    
    # Check if product is within the threshold
    if product <= 1000:
        return product
    else:
        return num1 + num2

# Testing Case 1
result = mult_or_sum(20, 30)
print("The result is", result)

# Testing Case 2
result = mult_or_sum(40, 30)
print("The result is", result)
```

<p style="font-size:24px; font-weight:bold;">16. Write a program to check if a given number is a palindrome (reads the same forwards and backwards).</p>

```python
def check_palindrome(number):
    print("original number", number)
    # Convert to string to reverse easily
    original_str = str(number)
    reversed_str = original_str[::-1]
    
    if original_str == reversed_str:
        print("Yes. given number is palindrome number")
    else:
        print("No. given number is not palindrome number")

check_palindrome(121)
check_palindrome(125)
```

<p style = "font-size:24px; font-weight:bold;">22. Write a function called <span style="background-color: #FF49Fa; padding: 2px 6px; border-radius: 4px; font-family: monospace;">exponent(base, exp)</span>  that returns an integer value of the base raised to the power of the exponent.</p>

```python
def exponent(base, exp):
    num = exp
    result = 1
    # Repeat multiplication 'exp' times
    while num > 0:
        result = result * base
        num = num - 1
    print(base, "raises to the power of", exp, "is:", result)

exponent(2, 5)
exponent(5, 4)
```

<p style = "font-size:24px; font-weight:bold;">25. Write a program that takes a year as input and determines if it is a leap year. </p>

```python
def is_leap(year):
    # Standard Leap Year Logic
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        print(f"{year} is a leap year")
    else:
        print(f"{year} is not a leap year")

is_leap(2024)
is_leap(2100)
is_leap(2027)
```

<p style = "font-size:24px; font-weight:bold;">31. Write a program to find all prime numbers up to 20, but only print every second (alternate) prime number found.</p>

```python
primes = []

for num in range(2, 21):
    # Check if number is prime
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            break
    else:
        primes.append(num)

# Print alternate primes
alternate_primes = primes[::2]
print(alternate_primes)
```

```python
class Car:
    def __init__(self, make, model, year):
        # Setting up attributes
        self.make = make
        self.model = model
        self.year = year

    def start_engine(self):
        # A method that uses the object's attributes
        print(f"The {self.year} {self.make} {self.model}'s engine is now running!")

# Creating an object (an instance of the class)
my_car = Car("Toyota", "Camry", 2022)

# Calling the method
my_car.start_engine()
```
