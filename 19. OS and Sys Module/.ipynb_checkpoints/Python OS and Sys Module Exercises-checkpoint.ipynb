<p style="font-weight:bold; font-size:24px"> 1. Write a Python program that prints the current working directory to the console.</p>

```python
import os

cwd = os.getcwd()
print("Current Directory:", cwd)
```

<p style="font-weight:bold; font-size:24px"> 2. Write a Python program that lists all files and folders in a given directory path.</p>

```python
import os

path = "."
contents = os.listdir(path)

for item in contents:
    print(item)
```

<p style="font-weight:bold; font-size:24px"> 3. Write a Python program that creates a new folder called <span style="background-color: #FFF0FF; padding: 2px 6px; border-radius: 4px;">test_folder </span> in the current working directory, only if it does not already exist.</p>

```python
import os

folder_name = "test_folder"

if not os.path.exists(folder_name):
    os.mkdir(folder_name)
    print(f"Folder '{folder_name}' created successfully.")
else:
    print(f"Folder '{folder_name}' already exists.")
```

<p style="font-weight:bold; font-size:24px"> 4. Write a Python program that creates a nested directory structure <span style="background-color: #FFF0FF; padding: 2px 6px; border-radius: 4px;"> a/b/c </span> in the current working directory, even if some or all of the parent directories do not yet exist.</p>

```python
import os

nested_path = os.path.join("a", "b", "c")

os.makedirs(nested_path, exist_ok=True)
print(f"Nested directories '{nested_path}' created successfully.")
```

<p style="font-weight:bold; font-size:24px"> 5.  Write a Python program that renames a file called <span style="background-color: #FFF0FF; padding: 2px 6px; border-radius: 4px;"> old.txt </span> to <span style="background-color: #FFF0FF; padding: 2px 6px; border-radius: 4px;"> new.txt </span> in the current directory, with a check to confirm the source file exists before attempting the rename.</p>

```python
import os

src = "old.txt"
dst = "new.txt"

# Create the file for testing purposes
open(src, "w").close()

if os.path.exists(src):
    os.rename(src, dst)
    print(f"Renamed '{src}' to '{dst}' successfully.")
else:
    print(f"Source file '{src}' does not exist.")
```
